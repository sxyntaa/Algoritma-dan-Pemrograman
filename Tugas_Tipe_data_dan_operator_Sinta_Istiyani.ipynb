**1. Membuat satu contoh dari masing-masing tipe data**

```python
#Tipe data list
ls = ["satu", "dua", "tiga", "empat"]
print(ls)
```

```python
#Tipe data tuple
tupl = ("satu", "dua", "tiga", "empat")
print(tupl)
```

2. **Buat rumus luas lingkaran menggunakan phi 3.14**

```python
#Menentukan luas lingkaran
phi = 3.14
r = 18
luas = phi * (r ** 2)
print("luas lingkaran:", luas)
```

**3. Total belanja**

```python
#Total belanja pada sebuah toko
Buku = 3000
Pensil = 1000
Penghapus = 500

Total_agung = (3 * Buku) + (2 * Pensil)
Total_sandy = (2 * Buku) + (4 * Penghapus)
Total_semua = Total_agung + Total_sandy

print("Total belanja agung:", Total_agung)
print("Total belanja sandy:", Total_sandy)
print("Total Keseluruhan:", Total_semua)
```
